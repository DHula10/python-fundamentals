## 1. Calculations & Modul Math
Ukázka práce se zápisem velkých celých čísel a matematickými funkcemi ze standardní knihovny `math`.

```python

#* Podtržítka v číslech slouží jako oddělovače tisíců pro lepší čitelnost (Python je ignoruje)
first_arithmetics = 4 - 267_865_448 + 534_403

import math

#? log10 počítá dekadický logaritmus výrazu ((4 + 4) / 5)
log_calculation = math.log10((4 + 4) / 5)

print("first_arithmetics:", first_arithmetics)
print("log_calculation:", log_calculation)
```

## 2. Variable Usage & Data Types
Přiřazení skalárních typů (`int`, `float`), n-tic (`tuple`), seznamů (`list`), generování psudonáhodných čísel a slovníků (`dict`).

```python

#* Skalární proměnné
number4 = 4
float45 = 4.5

#! N-tice (tuple) je neměnná (immutable) kolekce definovaná v kulatých závorkách
small_tuple = (0, 1, 2, 3, 4, 5)

#* Seznam (list) může obsahovat prvky různých datových typů i vnořené kolekce
list_for_all = [number4, float45, small_tuple]

import random
#! Nastavení seedu zaručuje reprodukovatelnost pseudonáhodných čísel
random.seed(42)
random_1000 = random.randint(0, 1000)

#? Slovník (dict) ukládá páry klíč: hodnota
book = {
    "nazev": "Malý princ",
    "autor": "Antoine de Saint-Exupéry",
    "rok_vydani": 1943,
    "cena": 250,
    "skladem": True,
    "zanry": ["pohádka", "filozofie", "klasika"]
}

print("list_for_all:", list_for_all)
print("random_1000:", random_1000)
print("book:", book)
```

## 3. Work with Lists & Slicing
Generování sekvencí pomocí `range()`, zjištění délky, indexování od začátku i od konce a řezání (slicing) s krokem.

```python

#? range(100) generuje čísla 0 až 99 (horní mez je vždy exkluzivní)
first100 = list(range(100))
len_of_list = len(first100)

#* Indexování prvků:
first_item = first100[0]

#! Záporný index -1 vybírá poslední prvek od konce
last_item = first100[-1]

#* Slicing syntaxe: [start : stop : step]
#? Lichá čísla: od indexu 1, krok 2
odd_numbers = first100[1::2]

#? Sudá čísla: od indexu 0, krok 2
even_numbers = first100[::2]

#! Krok -1 obrátí pořadí celého seznamu bez nutnosti volat list.reverse()
reversed_first100 = first100[::-1]

print("Délka seznamu:", len_of_list)
print("První a poslední prvek:", first_item, ",", last_item)
print("Prvních 5 lichých čísel:", odd_numbers[:5])
print("Prvních 5 sudých čísel:", even_numbers[:5])
print("Prvních 5 obrácených čísel:", reversed_first100[:5])
```

## 4. Work with Tuples
Vytvoření neměnné sekvence (tuple) pomocí konstruktoru `tuple()`

```python

#* Konstruktor tuple() vytvoří n-tici z libovolného iterovatelného objektu
tuple_first100 = tuple(range(100))
len_of_tuple = len(tuple_first100)

print("len_of_tuple:", len_of_tuple)
print("Ukázka n-tice (prvních 5):", tuple_first100[:5])
```

## 5. Control Flow & Loops
Iterace přes seznam pomocí cyklu `for`, vyhodnocení podmínek s modulo operátorem `%` a plnění výsledného seznamu.

```python
n_list = [2, 3, 6, 7, 8, 9, 10, 11, 17, 28, 37]
modulo_02 = []

#* Procházení prvků a filtrování podle zbytku po dělení třemi
for x in n_list:
    #? Podmínka zachytí čísla, kde zbytek po dělení 3 je roven 0 nebo 2
    if x % 3 == 0 or x % 3 == 2:
        modulo_02.append(x)

print("Původní seznam:", n_list)
print("Filtrovaný seznam (modulo 0 nebo 2 po dělení 3):", modulo_02)
```

## 6. Functions
Definice vlastní funkce s parametrem a návratovou hodnotou (`return`).

```python
import math

#* Definice funkce pro výpočet obsahu kruhu: S = π * r²
def circle_area(r):
    """Vypočítá obsah kruhu pro zadaný poloměr r."""
    #! Operátor ** slouží v Pythonu k umocňování
    return math.pi * (r ** 2)

#? Otestování funkce pro poloměr r = 10
area = circle_area(10)

print(f"Obsah kruhu s poloměrem 10 je: {area: .4f}")
```
